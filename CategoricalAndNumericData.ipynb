```python
exam_data = pd.read_csv("exams.csv", quotechar='"')
exam_data

# The data appears to be out of 100, but it is possible that the range of scores is different, therefore possibly making them not directly comparable
# This makes standardization essential
```

```python
math_average = exam_data['math score'].mean()
reading_average = exam_data['reading score'].mean()
writing_average = exam_data['writing score'].mean()

print('Math Avg: ', math_average)
print('Reading Avg: ', reading_average)
print('Writing Avg: ', writing_average)
```

```python
# Skikit-learn offers a very useful toolkit called preprocessing, which can be used to standardize a dataset
# The SCALE function will divide all the data points, for example the math scores by the average and subtract the standard deviation

from sklearn import preprocessing

exam_data[['math score']] = preprocessing.scale(exam_data[['math score']])
exam_data[['reading score']] = preprocessing.scale(exam_data[['reading score']])
exam_data[['writing score']] = preprocessing.scale(exam_data[['writing score']])

# To check the updated dataset with standardized numbers
exam_data
```

```python
math_average = exam_data['math score'].mean()
reading_average = exam_data['reading score'].mean()
writing_average = exam_data['writing score'].mean()

print('Math Avg: ', math_average)
print('Reading Avg: ', reading_average)
print('Writing Avg: ', writing_average)

# Checking the averages of these columns again. The averages are very close or equal to zero because the mean and stdev have been subtracted
# After scaling, the mean is zero (or almost zero) and the stdev of the scores is one
```

```python
# A very useful library function that scikit-learn offers is the label encoder
# It is used when there are meaningful comparisions between the categorical variables e.g. low<medium<high
# Only meaningful to the developer, not the machine learning model

le = preprocessing.LabelEncoder()
exam_data['gender'] = le.fit_transform(exam_data['gender'].astype(str)) 
# the fit_transform method will go through all the categorical values in the gender column and transform them into numbers. The numerical values for
# gender will be assigned to the gender column in the pandas data frame
```
