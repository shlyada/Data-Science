```python
import pandas as pd # To read in and explore the data
```

```python
auto_data = pd.read_csv('imports-85.data', sep=r'\s*,\s*', engine='python') 
# The sep=r'\s*,\s* is a passed argument in order to seperate the data with commas
# The python engine within pandas is being used to read the file
auto_data
```

```python
import numpy as np

auto_data = auto_data.replace('?', np.nan) # NAN means "Not A Number". The ?'s are being replaced by NANs
auto_data.head()
```

```python
auto_data.describe() # The "price" field is missing here
```

```python
auto_data.describe(include='all') # To get all the data
```

```python
auto_data['price'].describe()
# The dtype is 'object', which means it cannot be summarized with numerical statistics
```

```python
auto_data['price'] = pd.to_numeric(auto_data['price'], errors='coerce')
# pd.to_numeric converts this column into numerical values; errors is set to coerce, meaning we want to ignore the errors and force the conversion
```

```python
auto_data['price'].describe() # Now this column's dtype is float64
```

```python
auto_data = auto_data.drop('normalized-losses', axis=1) # this column is being dropped as it has no effect on the target variable: price
auto_data.head()
```

```python
auto_data['price'].describe()
```

```python
auto_data['horsepower'].describe()
```

```python
auto_data['horsepower'] = pd.to_numeric(auto_data['horsepower'], errors='coerce')
```

```python
auto_data['horsepower'].describe()
```

```python
auto_data['num-of-cylinders'].describe()
```

```python
cylinders_dict = {'two': 2, 'three': 3, 'four': 4, 'five': 5, 'six': 6, 'eight': 8, 'twelve': 12}
auto_data['num-of-cylinders'].replace(cylinders_dict, inplace=True)

auto_data.head()
```

```python
auto_data['num-of-cylinders'].describe()
```

```python
auto_data = pd.get_dummies(auto_data, columns=['make', 
                                               'fuel-type', 
                                               'aspiration', 
                                               'num-of-doors', 
                                               'body-style', 
                                               'drive-wheels', 
                                               'engine-location', 
                                               'engine-type', 
                                               'fuel-system'], dtype='int')
auto_data.head()
```

```python
auto_data = auto_data.dropna() # This drops all rows which don't have number values
auto_data
```

```python
auto_data[auto_data.isnull().any(axis=1)] # To check whether the dataset has any null values
```

```python
from sklearn.model_selection import train_test_split

X = auto_data.drop('price', axis=1)

# Taking the labels/target variable (price)
Y = auto_data['price']

# Splitting into 80% for training set and 20% for testing set so we can see the accuracy
X_train, x_test, Y_train, y_test = train_test_split(X, Y, test_size = 0.2, random_state = 0)
```

```python
from sklearn.linear_model import LinearRegression

linear_model = LinearRegression() # Instantiate a linear regression object called an Estimator
# An Estimator is a high level API that scikit-learn offers which implements the fit() method. 
# Any object in scikit-learn which learns from data which can be trained is an estimator

linear_model.fit(X_train, Y_train) # Passing in the X data and Y variables to start the training process for the estimator

# Should look something like: 
# *LinearRegression(copy_X=True, fit_intercept=True, n_jobs=1/None, normalize=False)* - There are the DEFAULT VALUES
# copy_X=True meanings that X is copied and not overwritten
# fit_intercept is set to True to indicate that intercepts are being used. If they weren't, data would not be centered around zero
# normalize is set to False indicating that we don't want the data to be normalized. 
# Normalization involves subtracting the mean and dividing by the L2 norm
```

```python
linear_model.score(X_train, Y_train) # This is in order to get the R squared of this regression model, by calling the score method on the linear model
```

```python
linear_model.coef_ # The coefficients on the linear model; the output of this shows the weights for individual features
```

```python
predictors = X_train.columns # To get the feature names from the columns of the training data set
coef = pd.Series(linear_model.coef_,predictors).sort_values() 
# Settng up a Pandas series that will associate the coeffs with the feature names and sort them by coeffs
# 'linear_model.coef_,predictors' allows to associate each predictor with the feature that it corresponds to
# This is to see how much weight is given to a particular feature in the final regression model
print(coef)
```
