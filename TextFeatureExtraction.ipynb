```python
# Basic frequency-based representation of words in documents
from sklearn.feature_extraction.text import CountVectorizer
```

```python
# The corpus has 4 documents
corpus = ['This is the first document.','This is the second document.','Third document. Document number three','Number four. To repeat, number four']
```

```python
# initialize
vectorizer = CountVectorizer()
# generating a bag of words in a numeric form
bag_of_words = vectorizer.fit_transform(corpus)

bag_of_words
```

```python
print(bag_of_words) # format = (documentID, wordID)
# There are 4 documents in the input corpus, and therefore, every document is given a unique ID from 0 to 3, as the first coordinate in the points
# Every word in the document corpus is also given a unique individual ID
# The second coordinates in the first 5 points represent the first 5 words in document 1, and so on
# Both word frequencies and integer representations are found within this dataset

# An example of word frequency is the number 2 in the second column for the point (3,2), as the 4th document has the word 'four' repeat twice

# An example of integer representations includes the fact that because both the first and second documents contain all the same words except one, where
# the first doc says 'first' doc while the second doc says 'second' doc. This is represented down below, as otherwise the rest of the first 10 points
# are identical
```

```python
# In order to access the ID that corresponds to a particular word. In this case, the word 'document' corresponds to the number 0
vectorizer.vocabulary_.get('document')

# Typically the most frequently used words are assigned lower IDs
```

```python
vectorizer.vocabulary_ # in order to access all the IDs
```

```python
# To understand the set up of these words in a tabular format

pd.DataFrame(bag_of_words.toarray(), columns=vectorizer.get_feature_names_out()) 

# rows are the individual documents, columns are the words in the vocab
# the numbers in the cells of the table represent the frequency of individual words in each doc
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer() # associates scores with every word in the document corpus
bag_of_words = vectorizer.fit_transform(corpus)
```

```python
print(bag_of_words) # all the document ID word ID combinations are associated with a score as seen below
```

```python
from sklearn.feature_extraction.text import HashingVectorizer

vectorizer = HashingVectorizer(n_features=8) # use Hashing instead of Count if there is a large vocabulary amount of words
# The use of hashing buckets to represent words allows to scale large data sets; n_features is the input argument for the number of hash buckets
feature_vector = vectorizer.fit_transform(corpus)
print(feature_vector)

# Word IDs are from 0-7 (the second coordinates of the points), as there are 8 buckets
# Due to the size of the vocab being larger than the # of buckets, as it should be, multiple words can hash to the same bucket
# A disadvantage of HashingVectorizer is that there is no way to get back to the original word from its hash bucket value
# The frequencies of each token is not represented in raw # form, as this is some kind of a normalized form
```
